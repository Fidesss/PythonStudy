1. 첫째 줄에 행렬의 크기가 주어짐.
2. 둘째 줄부터 N개의 줄에 행렬 A의 원소 M개가 주어짐
* 행렬의 합 > 같은 위치에 있는 리스트에서 원소를 하나씩 빼서 각각 더하기

### 문제풀이
* map함수를 사용하여 입력값을 공백을 기준으로 나눈 후 변수에 저장
* 리스트에 요소를 추가하는 append함수를 사용할 때 map함수를 같이 사용
* 이중 반복문으로 위치에 맞는 숫자들을 더한 후 출력

```python
a, b = map(int, input().split())

list_a = []
list_b = []

for _ in range(a):
    list_a.append(list(map(int, input().split())))

for _ in range(a):
    list_b.append(list(map(int, input().split())))

for i in range(a):
    for j in range(b):
        print(list_a[i][j] + list_b[i][j], end=' ')
    print()
```
