### 다이얼 문제
* 다이얼은 숫자 하나를 금속판까지 돌리면 다시 제자리로감
* 1을 돌리면 2초가 걸리면 옆으로 1칸 이동할 때 마다 1초가 더 걸린다.
* UNUCIC > 868242, 
8 > U, 6 > N, 2 > C, 4 > I

## 노트
2~9까지 각 알파벳이 대응 되는 알파벳을 리스트로 가지고있다가 거기 있으면 확인하는 식으로 하면 될 것 같은데
* 2 : ABC
* 3 : DEF
* 4 : GHI
* 5 : JKL
* 6 : MNO
* 7 : PQRS
* 8 : TUV
* 9 : WXYZ

```python
input_a = input()

dial = ["ABC", "DEF", "GHI", "JKL", "MNO", "PQRS", "TUV", "WXYZ"]

total = 0
for x in input_a:
    for i, j in enumerate(dial):
        if x in j:
            total += i+2

print(total)
```
