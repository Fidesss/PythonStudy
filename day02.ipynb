### input함수로 받은 값 캐스팅하기

```python
str_a = input("입력A> ")
print("입력A의 타입",type(str_a))
int_a = int(str_a)
print("캐스팅 후 타입 : ", type(int_a))

str_b = input("입력B> ")
print("입력B의 타입",type(str_b))
int_b = int(str_b)
print("캐스팅 후 타입 : ", type(int_b))

print("문자열 자료:", str_a + str_b)
print("숫자 자료:", int_a + int_b)
```

```python
# format() 함수로 숫자를 문자열로 변환하기
string_a = "{}".format(10)
print(string_a)
print(type(string_a))
```

```python
# {:d} << 이거는 정수만 출력한다 즉 format함수에 정수말고 
# 다른 자료형이 들어가면 안된다는 것을 의미한다.


output_a = "{:d}".format(52)
print(output_a)
# print(type(output_a))

output_b = "{:5d}".format(52)
print(output_b)
# print(type(output_b))

output_c = "{:10d}".format(52)
print(output_c)

output_d = "{:05d}".format(52)
print(output_d)

output_e = "{:010d}".format(52)
print(output_e)
```

```python
output_h = "{:=+05d}".format(52)
print(output_h)
```

```python
output_i = "{:d}".format(52)
print(output_i)
```

```python
output_z = "{:+1d}".format(52)
print(output_z)
```

```python
output_a = "{:.100f}".format(52.273)
print(output_a)
output_b = "{:15f}".format(52.273)
print(output_b)
output_c = "{:15.3f}".format(52.273)
print(output_c)
```

```python
# 부동소수점 오차
a = 0.1 + 0.2
b = 0.3
if a == b:
    print("True")
else:
    print("False")


o01 = "{:.100f}".format(a)
o02 = "{:.100f}".format(b)


print(o01)
print(o02)
```

```python
output_ca1 = "{:+>10d}".format(53)
output_ca2 = "{:+<10d}".format(53)
output_ca3 = "{:+^10d}".format(53)
output_cb = "{:+>10d}".format(-53)
output_cc = "{:+>+10d}".format(-53)
print(output_ca1)
print(output_ca2)
print(output_ca3)
print(output_cb)
print(output_cc)
```

```python
#:g를 사용하면 파이썬이 생각했을 때 가장 깔끔한 코드로 변경해준다.

output_a = 52.0
output_b = "{:g}".format(output_a)
print(output_a)
print(output_b)
```
