```python
# 예제 : 직사각형 넓이와 둘레 구하기

width = 8
height = 5

print("가로 =", width)
print("세로 =", height)
print("직사각형의 둘레 =", (width * 2) + (height * 2))
print("직사각형의 넓이 =", width * height)
```
