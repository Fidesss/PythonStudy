```python
try:
    num = input("정수 입력> ")
    num = int(num)
    if num > 0:
        print("양수입니다.")
    elif num < 0:
        print("음수 입니다.")
    else:
        print("0 입니다.")
except ZeroDivisionError:
    print("0으로 나눌 수 없습니다.")
except ValueError:
    print("올바른 정수를 입력하세요.")
except TypeError:
    print("올바른 자료형을 입력하세요.")
```
