```python
name = input("이름> ")
phone = input("전화번호> ")
price = input("주문 금액> ")

# name = "윤땡땡"
# phone = "01012345678"
# price = "15000"

order = int(price)
discount = int(order/10)
total = order - discount

print("이름: "+ name)
print("전화번호 앞자리: "+ phone[:3])
print("전화번호 뒷자리: "+ phone[7:])
print("사용자 ID: "+ name + phone[7:])
print("주문 금액: "+ price +"원")
print("할인 금액: "+ str(discount) +"원")
print("최종 결제 금액: "+ str(total) +"원")
```
