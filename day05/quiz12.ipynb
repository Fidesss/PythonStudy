### 21번

```python
ohlc = [["open", "high", "low", "close"],
    [100, 110, 70, 100],
    [200, 210, 180, 190],
    [300, 310, 300, 310]]

# 3일간의 ohlc 데이터 > 종가가 시가보다 높은 날의 변동성 (고가 - 저가)를 출력
# 일 > index, 각 인덱스에 있는 값을 비교
# close > open

for idx, val in enumerate(ohlc):
    if idx == 0:
        continue
    if ohlc[idx][3] > ohlc[idx][0]:
        print(f"{idx}일 변동성은 {ohlc[idx][1] - ohlc[idx][2]}")
```

### 27번

```python
cart = [
    {"name": "우유", "price": 2500, "qty": 2},
    {"name": "커피", "price": 3000, "qty": 1},
    {"name": "빵", "price": 2000, "qty": 3},
]

total_price = 0

print("실행결과:")
print("장바구니 내역")

# cart 리스트의 요소인 딕셔너리의 key 값 name, price, qty는 고정
for idx, item in enumerate(cart):
    print(f"{cart[idx]["name"]} {cart[idx]["qty"]}개 {cart[idx]["price"] * cart[idx]["qty"]}원")
    total_price += cart[idx]["price"] * cart[idx]["qty"]
print(f"총 결제 금액: {total_price}원")
```
