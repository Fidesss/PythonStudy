## 1번 문제

```python
# 짝수인데 10보다 큰 숫자만 제곱하여 리스트 만들기
numbers = [3, 12, 5, 20, 7, 18, 1]
arr_a = [i*i for i in numbers if i % 2 == 0 and i > 10]
print(arr_a)
```

## 2번 문제

```python
# 이름 길이가 4글자 이상인 사람만 대문자 변환
names = ["anna", "tom", "charlie", "bob", "julia"]

# if else문을 사용하려면 if문을 for 앞으로 옮겨야 한다.
names = [i.upper() if len(i) >= 4 else i for i in names]
print(names)
```

## 3번 문제

```python
# 모음 제거하기 a e i o u
words = ["apple", "banana", "kiwi", "grape"]

temp = ""

for idx, i in enumerate(words):
    for j in i:
        if j not in "aeiou":
            temp += j
    words[idx] = temp
    temp = ""

print(words)
```

```python
# 모음 제거하기 a e i o u
# join()메서드는 붙이는 역할을 한다.

words = ["apple", "banana", "kiwi", "grape"]

a = ["".join([j for j in i if j not in "aeiou"]) for i in words ]

print(a)
```

#### 생각 흐름

```python
a = ["".join(j) for i in words for j in i if j not in "aeiou"]

print(a)
```

```python
a = [ [j for j in i if j not in "aeiou"] for i in words]

print(a)
```

```python
a = ["".join([j for j in i if j not in "aeiou"]) for i in words ]

print(a)
```

## 4번 문제

```python
scores = {
    "민수" : 85,
    "지영" : 42,
    "태리" : 91,
    "하리" : 58,
    "수빈" : 77
}

passedUsers = {}

for key, val in scores.items():
    if val >= 60:
        passedUsers.update({key: val})

print(passedUsers)
```

```python
scores = {
    "민수" : 85,
    "지영" : 42,
    "태리" : 91,
    "하리" : 58,
    "수빈" : 77
}

passedUsers = {key: val for key, val in scores.items() if val >= 60} 

print(passedUsers)
```
