* reversed() 함수 : 요소를 반전시킨다.

```python
list_a = [1, 2, 3, 4, 5]
list_reversed = reversed(list_a)

# 출력합니다.
print("# reversed() 함수")
print("reversed([1, 2, 3, 4, 5])", list_reversed)
print("list(reversed([1, 2, 3, 4, 5])):", list(list_reversed))
print()

# 반복문을 적용해 봅니다.
print("# reversed() 함수와 반복문")
print("for i in reversed([1, 2, 3, 4, 5]):")
for i in reversed(list_a):
    print("-", i)
```

* reversed() 함수와 반복문 조합할 때는 함수 결과 여러 번 활용하지 않고 for 구문 내부에 reversed() 함수 곧바로 사용
* 하나의 리스트에 reversed()함수를 사용하면 이터레이터(한 번 쓰면 사라짐)이기 때문에 변수에 넣어서 사용하는 것 보다 for문 내에 reversed() 함수를 사용하여 여러번 사용할 수 있게 하기
* 아니면 그냥 list로 만들어서 하면 되지 않을까 싶기도 함 이건 테스트 해보기 > 된다.

```python
a = [1, 2, 3, 4, 5]

temp = list(reversed(a))
print(type(temp))
print(temp)

for i in temp:
    print(i)
print("-------------------------")

for i in temp:
    print(i)
```

* enumerate() 함수 > 리스트의 인덱스 값과 요소 값이 같이 나온다. item()함수와 비슷
* enumerate는 (순번, 그 객체를 for문으로 돌렸을 때 나오는 값) 즉 딕셔너리에서 사용을 해도 키, 밸류 이런식으로 나눠서 나오지는 않는다.
* 딕셔너리에서 키, 밸류 이런식으로 사용하려면 items() 함수를 사용하면 된다.

```python
example_list = ["요소A", "요소B", "요소C"]
example_dict = {"요소A": "A", "요소B": "B", "요소C": "C"}


for idx, val in enumerate(example_list):
    if idx == 0:
        print("리스트")
        print("인덱스   요소")
    print(f"{idx}\t{val}")

print()

for key, val in enumerate(example_dict):
    if key == 0:
        print("딕셔너리")
        print("키   값")
    print(f"{key}\t{val}")
```

```python
print("딕셔너리")
print("키   값")
for key, val in example_dict.items():
    print(f"{key}\t{val}")
```

만약 순번 까지 필요하다면?

```python
for idx, (key, val) in enumerate(example_dict.items()):
    if idx == 0:
        print("딕셔너리")
        print("인덱스   키   값")
    print(f"{idx}\t{key}\t{val}")
```

```python
# 최댓값 문제처럼 번호가 1부터 시작할 때 따로 +1을 하지 않아도 된다.
# enumerate함수에 파라미터로 start = 값 > 이런식으로 추가해주면 된다.
for idx, val in enumerate(example_list, start=1):
    print(idx, val)   # 1부터 시작
```

```python
example_list = ["요소A", "요소B", "요소C"]
print("리스트로 강제 변환 출력")
print(list(enumerate(example_list)))

print("map함수를 사용하여 튜플을 리스트로 변경")
print(list(map(list, enumerate(example_list))))
```
