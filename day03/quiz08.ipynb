```python
# 문제 1 숫자 1~5까지 출력하기
for i in range(5):
    print(i + 1)
```

```python
# 문제 2 : 다음 리스트 안의 과일을 하나씩 출력하기 ["사과","바나나","딸기"]
fruits = ["사과", "바나나", "딸기"]
for i in fruits:
    print(i)
```

```python
# 문제 3 리스트 숫자들의 합 구하기
numbers = [10, 20, 30]
total = 0
for i in numbers:
    total += i
print(total)
```
