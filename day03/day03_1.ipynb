### if 조건문

```python
# 날짜/시간 출력하기
import datetime
import time

now = datetime.datetime.now()

a = 0

# print(type(now))

while a < 3:
    a += 1
    if a == 1:
        print("=====================")
    print(
        "{}년 {}월 {}일 {}시 {}분 {}초".format(
            now.year, now.month, now.day, now.hour, now.minute, now.second
        )
    )
    print("=====================")
    time.sleep(2)
```

```python
print("{}년 {}월 {}일 {}시 {}분 {}초".format(
    now.year,
    now.month,
    now.day,
    now.hour,
    now.minute,
    now.second
))
```

```python
# 오전 오후 구분 프로그램

if now.hour < 12:
    print("현재 시간은 {}시 이고 오전 입니다.".format(now.hour))
else:
    print("현재 시간은 {}시 이고 오후 입니다.".format(now.hour))
```

```python
#예시 숫자를 입력받아 끝자리로 짝수와 홀수를 구분하는 프로그램을 만들어라.
num = input("정수 입력> ")

# 마지막 자리 추출
last_char = num[-1]
last_num = int(last_char)

# 조건문
if (last_num % 2) == 0:
    print("짝수 입니다.")
else:
    print("홀수 입니다.")
```

```python
## 계절 3~5 > 봄, 6~8 > 여름, 9~11 > 가을 나머지 겨울
month = now.month

if 3 <= month <= 5:
    print("spring")
elif 6 <= month <= 8:
    print("summer")
elif 9 <= month <= 11:
    print("fall")
else:
    print("winter")
```

```python
# 학점 계산 프로그램
score = float(input("학점 입력> "))

if score == 4.5:
    print("a")
elif 4.2 <= score:
    print("b")
elif 3.5 <= score:
    print("c")
elif 2.8 <= score:
    print("d")
elif 2.3 <= score:
    print("e")
elif 1.75 <= score:
    print("f")
elif 1.0 <= score:
    print("g")
elif 0.5 <= score:
    print("h")
elif 0 < score:
    print("i")
else:
    print("시대를 앞서가는 혁명의 씨앗")
```
