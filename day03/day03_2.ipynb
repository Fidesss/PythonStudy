### 리스트에 요소를 추가하는 메서드
* append() : 리스트 뒤에 요소를 추가
* insert() : 리스트 중간에 요소를 추가

```python
# 리스트 뒤에 요소 추가하기
list_a = [1, 2, 3]
print(list_a)

# 리스트 뒤에 요소 추가하기
print("# 리스트 뒤에 요소 추가하기")
list_a.append(4)
list_a.append(5)
print(list_a)
print()

print("# 리스트 중간에 요소 추가하기")
list_a.insert(0, 10)
print(list_a)
```

* extend() 함수

```python
list_a = [1, 2, 3]
list_a.extend([4, 5, 6])
print(list_a)
```

* 다시 저장하지 않고 + - 연산을 사용하여 리스트에 요소를 추가하거나 삭제할 경우 원본에는 지장이 없음.
* append, insert, extend와 같은 메서드를 사용하여 요소를 추가할 경우 원본에 영향을 미친다.

```python
list_a = [1,2,3]
list_b = [6,7,8]

print(list_a + list_b)
print(list_a)
print(list_b)

list_a.append("asdfsdf")
print(list_a)


list_a.extend(list_b)
print(list_a)

list_a.insert(0, list_b)
print(list_a)
```

## 리스트에서 요소 제거하기
* 제거방법 1 : del
* 제거방법 2 : pop

```python
list_a = [0, 1, 2, 3, 4, 5]
print("# 리스트의 요소 하나 제거하기")

# 제거방법 1, - del
del list_a[1]
print("del list_a[1]:", list_a)

# 제거방법 2, - pop
x1 = list_a.pop(2)
print("pop is :", x1)
print("pop(2):", list_a)
```

```python
list_b = [0,1,2,3,4,5]
del list_b[:4]
list_b
```

* remove() : 특정 값을 지정하여 제거
* 리스트의 첫 번째 요소부터 차례대로 검사하여 특정 값을 제거하면 뒤 요소에 똑같은 특정 값이 있더라고 remove메서드의 할 일은 끝난다.
* 만약 없는 값이 들어가면 ValueError 발생

```python
list_c = [1,2,2,2,1]
list_c.remove(2)
list_c
```

```python
list_c = [1,2,2,2,1]
list_c.remove(3)
list_c
```

* 모두 제거하기 claer() 메서드
* 리스트 자체는 살아있지만 리스트 안의 모든 요소들이 제거된다.

### 확인 문제 2
#### 다음 반복문 내부에 if 조건문의 조건식을 채워서 100이상의 숫자만 출력하게 하라.

```python
numbers = [273, 103, 5, 32, 65, 9, 72, 800, 99]

for number in numbers:
    if number >= 100:
        print("- 100 이상의 수:", number)
```
