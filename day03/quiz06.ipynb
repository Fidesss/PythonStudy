```python
full_input = input('"이름 근무시간" 형태로 입력해주세요> ')

# 이름과 숫자를 구분하기 위해 공백을 기준으로 슬라이싱
edit_pos = full_input.find(" ")
name = full_input[:edit_pos]
work_hour = int(full_input[edit_pos + 1 :])

pay = 10000  # 시급
total_pay = 0  # 총 급여

# 하루 근무 시간 8시간 초과 시, 초과분은 1.5배 시급으로 계산

if work_hour > 8:
    total_pay = int(((work_hour - 8) * pay * 1.5) + pay * 8)
else:
    total_pay = int(work_hour * pay)

print("{}님의 급여는 {}원 입니다.".format(name, total_pay))
```
