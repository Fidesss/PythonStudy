```python
import datetime as dt

name, subject = input('"이름 학습주제" 형식으로 입력해주세요> ').split()

now = dt.datetime.now()
today = now.date()
today_weekday_num = now.weekday()
today_list = ['월요일','화요일','수요일','목요일','금요일','토요일','일요일']


# now.weekday() 0~6 > 월~일
today_weekday = today_list[today_weekday_num]

print("{}_{}_{}_{}공부.md".format(name, today, today_weekday, subject))
```

```python
import datetime as dt

input_a = input("기념일 날짜를 입력해주세요> ")

today_day = dt.date.today()

d_day = dt.datetime.strptime(input_a, "%Y-%m-%d").date()

delta = (d_day - today_day).days

if delta == 0:
    print("오늘이 기념일 입니다.")
elif delta < 0:
    print("기념일이 {}일 지났습니다.".format(abs(delta)))
else:
    print("기념일이 {}일 남았습니다.".format(delta))
```

* 위의 문제에서 datetime.date -> 라는 객체
* d_day를 today_day와 같은 객체로 만들어주기 위한 date()메서드 사용
* delta가 음수일 경우 지남, 양수일 경우 기념일이 @일 남음 과 같은 로직
