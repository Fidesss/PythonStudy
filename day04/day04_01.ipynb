### 딕셔너리

```python
a = {"키A": 10, "키B": 20, "키C": 30, 1: 40}
```

```python
dictionary = {
    "name" : "7D 건조 망고",
    "type" : "7D 건조 망고",
    "ingredient" : ["망고", "설탕", "메타중아황산나트륨", "치자황색소"],
    "origin" : "필리핀",
}
```

```python
print("name:", dictionary["name"])
print("type:", dictionary["type"])
print("ingredient:", dictionary["ingredient"])
print("origin:", dictionary["origin"])
print()

# 값을 변경합니다.
dictionary["name"] = "8D 건조 망고"
print("name:", dictionary["name"])
```

```python
for x in range(0,4):
    print(dictionary["ingredient"][x])
```

* dictionary에서 키 값은 문자열로 생성해야한다.
* 키 값은 변하면 안된다. > 바뀌지 않는 값을 사용한다.
* 인덱스와 같은 개념으로 접근해야하기 때문에 중복 값이면 안된다.
* 키 값은 의미 있는 값으로 사용 > 여러명이 협업하기 때문

딕셔너리에 값 추가하기/제거하기

```python
dictionary["price"] = 5000 # 딕셔너리에 새로운 키, 값 추가
dictionary
```

```python
dictionary["name"] = "8D 건조 파인애플"
dictionary
```

```python
dict_a = {"name" : "8D 건조 파인애플", "type" : "당절임"}

print("요소 제거 이전 :", dict_a)

del dict_a["name"]
del dict_a["type"]

print("요소 제거 이후 :", dict_a)
```

```python
dict_a = {"name" : "8D 건조 파인애플", "type" : "당절임"}

for x in dict_a:
    print(x, ":", dict_a[x])
```
