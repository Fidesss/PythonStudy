```python
inventory = {"메로나": [300, 20], "비비빅": [400, 3], "죠스바": [250, 100]}

print("메로나의 가격 {}원".format(inventory["메로나"][0]))
print("매로나의 재고 {}개".format(inventory["메로나"][1]))

inventory["월드콘"] = [500, 7]

key_list = []
value_list = []

for x in inventory:
    key_list.append(x)
    value_list.append(inventory[x])

print("key 값으로만 구성된 리스트 : ", key_list)
print("value 값으로만 구성된 리스트 : ", value_list)
```

```python
inventory = {"메로나": [300, 20], "비비빅": [400, 3], "죠스바": [250, 100]}

print("메로나의 가격 {}원".format(inventory["메로나"][0]))
print("매로나의 재고 {}개".format(inventory["메로나"][1]))

inventory["월드콘"] = [500, 7]

key_list = list(inventory.keys())
value_list = list(inventory.values())

print("key 값으로만 구성된 리스트 : ", key_list)
print("value 값으로만 구성된 리스트 : ", value_list)
```

```python
inventory = {"메로나": [300, 20], "비비빅": [400, 3], "죠스바": [250, 100]}

print("메로나의 가격 {}원".format(inventory["메로나"][0]))
print("매로나의 재고 {}개".format(inventory["메로나"][1]))

inventory["월드콘"] = [500, 7]

key, value = list(zip(*inventory.items()))
key_list = list(key)
value_list = list(value)

print("key 값으로만 구성된 리스트 : ", key_list)
print("value 값으로만 구성된 리스트 : ", value_list)
```
