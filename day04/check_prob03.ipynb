```python
for x in range(10):
    if (x + 1) % 2 == 0:
        print("{}는 짝수입니다.".format(x + 1))
    else:
        print("{}는 홀수입니다.".format(x + 1))
```

```python
total = 0

for x in range(1, 32):
    if x % 3 == 0:
        total += x
print("1~31까지의 배수의 합 = {}".format(total))
```

```python
for a in range(1, 10):
    if a % 2 == 0:
        print("---{}단---".format(a))
        for b in range(1, 10):
            print("{} x {} = {}".format(a, b, a * b))
```
